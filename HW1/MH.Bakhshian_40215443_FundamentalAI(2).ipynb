<div style="font-size: 14px; direction: ltr; text-align: left;">
The question asks why we square the errors and take their average, and why we do not simply use the average of the errors themselves.

Several reasons can be given in response:

1) The first and most important reason is differentiability. The squared function is continuous and differentiable at every point. This property is extremely important and essential for derivative-based optimization algorithms such as gradient descent.

2) The second reason is that larger errors receive a heavier penalty. In practice, by squaring the difference between the actual output and the predicted output, we make the penalty for large errors significantly greater. As a result, the model attempts to avoid highly inaccurate predictions and balance the overall error.

3) The third reason is convexity. This property guarantees that there is only one true minimum. Therefore, regardless of the starting point, an algorithm such as gradient descent can eventually converge to that minimum. If convexity is not preserved, the algorithm may fail to reach the correct solution.

4) Another important reason is to prevent positive and negative errors from canceling each other out.
</div>

<div style="font-size: 14px; direction: ltr; text-align: left;">
Linear Regression:

</div>
<div style="font-size: 14px; direction: ltr; text-align: left;">
    Here, we need to implement the prediction, cost, and optimization loop functions.
</div>


```python
def predict(X, w, b):
    # x : input data , w : weight , b : bias
    # y_hat  = wX + b
    y_hat = np.dot(X, w) + b
    return y_hat
```

```python
def compute_cost(X, y, w, b):
    m = len(y) # how many sample do we have?
    y_hat = predict(X, w, b)
    cost = [1/(2 * m)] * np.sum((y_hat - y) ** 2) # (predict - actual)
    return cost
```

<div style="font-size: 14px; direction: ltr; text-align: left;">
You might wonder why we divided by the number of samples, that is, m, rather than by 2m.
</div>
<div style="font-size: 14px; direction: ltr; text-align: left;">
The reason is to make the differentiation cleaner: differentiating a squared expression brings down a factor of 2, which then cancels out with the 2 in the denominator.
</div>
<div style="font-size: 14px; direction: ltr; text-align: left;">
$$
J(w,b) = \frac{1}{2m} \sum_{i=1}^{m} (\hat{y}_i - y_i)^2
$$

</div>


```python
def gradient_descent(X, y, w_init, b_init, alpha, epochs):
    # init w, b : initialize w, b
    # alpha = learning rate
    # epochs : learning loop
    w, b = w_init, b_init
    m = len(y)
    for i in range(epochs):
        y_hat = predict(X, w, b)
        dw = (1 / m) * np.dot(X.T, (y_hat - y))
        db = (1 / m) * np.sum(y_hat - y)
        w = w - alpha * dw
        b = b - alpha * db
    return w, b
        
```

$$
\frac{\partial J}{\partial w} = \frac{1}{m} X^T (\hat{y} - y)
$$

$$
\frac{\partial J}{\partial b} = \frac{1}{m} \sum_{i=1}^{m} (\hat{y}_i - y_i)
$$

$$
w := w - \alpha \frac{\partial J}{\partial w}
$$
$$
b := b - \alpha \frac{\partial J}{\partial b}
$$

```python
import sklearn
from sklearn.datasets import make_blobs
import numpy as np
X_p, y_p = make_blobs(n_samples = 200, centers = 2, n_features = 2, random_state = 42)
y_p = np.where(y_p == 0, -1, 1).reshape(-1, 1)

X_with_bias = np.c_[np.ones((X_p.shape[0], 1)), X_p]
#piade sazi moaadle normal ba ravabet jabr
w_lsp = np.linalg.inv(X_with_bias.T @ X_with_bias) @ X_with_bias.T @ y_p
```

<div style="font-size: 14px; direction: ltr; text-align: left;">

</div>
<div style="font-size: 14px; direction: ltr; text-align: left;">
Perceptron:

</div>
<div style="font-size: 14px; direction: ltr; text-align: left;">
   Here, we use the perceptron method, which updates the weights whenever a misclassification is detected.
</div>


```python
def perceptron_train(X, y, eta=0.1, epochs=10):
    # Initialize weights (including bias) with zeros
    n_samples, n_features = X.shape
    w = np.zeros(n_features + 1)
    X_b = np.c_[np.ones((n_samples, 1)), X]
    
    for _ in range(epochs):
        for i in range(n_samples):
            #  Check the condition for misclassification
            # Hint: y_i * (w * x_i) <= 0
         
            if y[i] * np.dot(w, X_b[i]) <= 0:
                
                # Update weights
                w = w + eta * (y[i] * X_b[i])
                
    return w

w_perceptron = perceptron_train(X_p, y_p)
```

In this section, you need to evaluate the performance of the trained perceptron model by calculating the confusion matrix. This matrix shows the number of correct and incorrect predictions for each class.
</div>


```python
from sklearn.metrics import confusion_matrix, classification_report
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np


# Generate predictions (sign of X_b * w)
X_p_b = np.c_[np.ones((X_p.shape[0], 1)), X_p]
y_pred = np.sign(X_p_b @ w_perceptron)

# Compute confusion matrix

cm = confusion_matrix(y_p, y_pred)

 
sns.heatmap(cm, annot=True, fmt='d', cmap='Oranges', xticklabels=['-1', '1'], 
            yticklabels=['-1', '1'])
plt.title("Perceptron Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()

print(classification_report(y_p, y_pred))
```

In this scenario, the number of samples is exactly equal to the number of features (both are 50). From a linear algebra perspective, we are solving a system of 50 equations with 50 unknowns. Instead of discovering a generalizable underlying pattern in the data, linear regression simply solves the equations and finds a line, plane, or hyperplane that passes exactly through all 50 points.
</div>

<div style="font-size: 14px; direction: ltr; text-align: left;">  
The model has effectively learned and memorized the training data, including the noise present in it. When the model encounters new data (test data) with a different noise distribution, its performance will be disastrous.

This issue is closely related to the concept of the curse of dimensionality.
</div>
