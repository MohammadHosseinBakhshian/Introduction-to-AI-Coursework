<h3 style="text-align: left;">
        Solution to Question 1 - Section 1.1 (Fuzzy Complement)
</h3>

<h4 style="text-align: left;">
        Part 1: The Concept of $\mu_{\overline{warm}}(x)$ in an Air Conditioning System
</h4>

<p style="text-align: left;">
        In fuzzy set theory, the membership function $\mu_{warm}(x)$ represents the degree of membership of variable $x$ (temperature) in the fuzzy set "warm air." The complement of this set, denoted by $\mu_{\overline{warm}}(x)$, expresses the degree to which the same temperature belongs to the set "not warm."
</p>

<p style="text-align: left;">
        From an engineering perspective in an air conditioning system, this complement logically represents the opposite state, indicating to what extent the air does <strong>not</strong> possess the characteristics of being warm, which in practice is interpreted as degrees of membership to the sets "cool air" or "cold air."
</p>

<h4 style="text-align: left;">
        Part 2: The Necessity of Axiom c2 for the Complement Function
</h4>

<p style="text-align: left;">
        Axiom c2, which is mathematically defined in the problem statement as:
</p>

<p style="text-align: center;">
        $$ \text{If } a < b \implies c(a) \ge c(b) $$
</p>

<p style="text-align: left;">
        expresses the condition of <strong>"monotonic non-increasing"</strong> for the fuzzy complement function. The necessity of this principle is rooted in preserving semantic integrity when extending classical logic to fuzzy logic. (We can also mention that this principle must hold because the existence of an extremum must be maintained.)
</p>

<p style="text-align: left;">
        In the essence of logical reasoning, proposition $A$ and its negation (i.e., $\sim A$) have a bidirectional relationship. When the truthfulness of a phenomenon increases (e.g., the air becomes warmer and $a$ increases), it is logically impossible for its falsity (not-warmness) to remain constant or increase. Axiom c2 mathematically guarantees that the fuzzy complement operator correctly simulates this inherent concept of "opposition and contrast" in the continuous space of $[0,1]$.
</p>

<h4 style="text-align: left;">
        Part 3: Logical Consequences and Problems if Axiom c2 is Violated in a Decision-Making System
</h4>

<p style="text-align: left;">
        Violation of condition c2 means that in some parts of the function, as the input value ($a$) increases, the complement output ($c(a)$) also increases (ascending behavior). To understand the catastrophic nature of this in practice, consider the air conditioning system scenario:
</p>

<p style="text-align: left;">
        Suppose the ambient temperature is rising sharply. The system sensors detect this temperature increase and the value of $\mu_{warm}(x)$ approaches $1$. If the monotonic non-increasing condition is violated, the system arrives at the contradictory conclusion that the degree of "not-warm" (coldness) is also increasing simultaneously with the air getting warmer!
</p>

<p style="text-align: left;">
        A fuzzy inference system typically includes rules such as the following:
</p>

<div style="font-size: 14px; text-align: left; padding-left: 30px;">
        <i>Rule 1:</i> <b>IF</b> Air is <b>Warm</b> <b>THEN</b> Increase Cooling Power. <br>
        <i>Rule 2:</i> <b>IF</b> Air is <b>NOT Warm</b> <b>THEN</b> Turn on Heater.
</div>

<p style="text-align: left;">
        With the violation of axiom c2, when the room becomes very hot, the degree of membership to "not warm" falsely increases. As a result, the controller system mistakenly fires the second rule with high strength. The system reasons that <i>"the air is getting very cold, so I should turn on the heater!"</i>
</p>

<p style="text-align: left;">
        This logical bug leads to the creation of a <strong>destructive positive feedback loop</strong>. A room that is warming up, instead of being cooled, is mistakenly heated further by the system's heater, and the stability of the entire control system is destroyed. This is why in the design of fuzzy controllers, axiom c2 is an indispensable requirement.
</p>

<h3 style="text-align: left;">
        Solution to Question 1 - Section 2.1 (Fuzzy Union and Intersection / S-norms and T-norms)
</h3>

<h4 style="text-align: left;">
        a) Practical Perspective and Behavior Towards Changes
</h4>

<p style="text-align: left;">
        The operator $t(a,b) = \min(a,b)$, known as the standard fuzzy intersection operator, has a fundamental weakness in control applications: <strong>"insensitivity to changes in the larger operand."</strong>
</p>

<p style="text-align: left;">
        <strong>Numerical example:</strong> Suppose two sensor signals (after fuzzification) produce values $a = 0.2$ and $b_1 = 0.8$. The intersection of these two using the $\min$ operator is:
</p>

<p style="text-align: center;">
        $$ \min(0.2, 0.8) = 0.2 $$
</p>

<p style="text-align: left;">
        Now suppose environmental conditions change and the second signal increases to $b_2 = 0.9$. The new intersection would be:
</p>

<p style="text-align: center;">
        $$ \min(0.2, 0.9) = 0.2 $$
</p>

<p style="text-align: left;">
        As observed, the output did not change at all! This means that in a fuzzy controller, as long as $b > a$, any fluctuation, disturbance, or change in variable $b$ is completely ignored.
</p>

<p style="text-align: left;">
        However, if we use the algebraic product operator, i.e., $t_{ap}(a,b) = ab$:
</p>

<p style="text-align: center;">
        $$ t_{ap}(0.2, 0.8) = 0.2 \times 0.8 = 0.16 $$
        $$ t_{ap}(0.2, 0.9) = 0.2 \times 0.9 = 0.18 $$
</p>

<p style="text-align: left;">
        <strong>Why is the $\min$ behavior undesirable?</strong> We are typically seeking to create a "smooth control surface" so that the system response is without jumps. The $\min$ operator creates flat regions (plateaus) on the control surface. This means that despite changes in the system state, the control command remains constant until the smaller variable changes! This behavior causes mechanical wear and step-like responses in actuators, whereas algebraic product, by considering all changes, produces a much smoother control surface.
</p>

<h4 style="text-align: left;">
        b) Theoretical Perspective
</h4>

<p style="text-align: left;">
        From the perspective of mathematics and machine learning, the need for other families of T-norms (such as algebraic product) is critical for at least two theoretical reasons:
</p>

<p style="text-align: left;">
        <strong>1. Differentiability:</strong> The function $\min(a,b)$ is not differentiable at points where $a = b$. In artificial intelligence and neuro-fuzzy systems (such as ANFIS), we need to use error backpropagation algorithms based on gradient descent for parameter tuning and network training. The non-differentiability of $\min$ poses a theoretical challenge for computing gradients. In contrast, the operator $t_{ap}(a,b) = ab$ is continuous and infinitely differentiable at all points ($\frac{\partial}{\partial a}(ab) = b$), which makes it highly ideal for optimization methods.
</p>

<p style="text-align: left;">
        <strong>2. Interactivity property:</strong> In set theory, the $\min$ operator is a <strong>non-interactive</strong> T-norm, meaning that no compensation or interaction occurs between the two truth values. However, the product operator is a <strong>strictly Archimedean T-norm</strong>. This theoretical structure guarantees that the output always depends on <i>all</i> inputs, and the combination of fuzzy concepts in a multivariate system is performed in a fully integrated and interconnected manner.
</p>

<h3 style="text-align: left;">
        Solution to Question 3.1 - Fuzzy Intersection (T-norms) and Its Bounds (Spam Filter Analysis)
</h3>

<h4 style="text-align: left;">
        a) Why is it Logical that the Output of Any T-norm Never Exceeds $\min(a,b)$? What Constraint Does This Upper Bound Impose on the Concept of "AND"?
</h4>

<p style="text-align: left;">
        In logic, the "AND" operator (intersection) seeks to examine the <i>simultaneous</i> occurrence of two conditions. In fuzzy space, when we want to measure the truth value of the proposition "signal $a$ <b>AND</b> signal $b$ are both true," logically this overall value cannot exceed the truth value of the weakest (lowest) condition.
</p>

<p style="text-align: left;">
        In the spam filter example, suppose the links in an email are suspicious with degree $0.4$ ($a = 0.4$) and the sender's address is suspicious with degree $0.9$ ($b = 0.9$). When we ask "how suspicious are the links <b>AND</b> the sender?" the answer cannot exceed $0.4$, because at least one of the conditions (the links) has only proven its suspiciousness to the degree of $0.4$.
</p>

<p style="text-align: left;">
        This upper bound imposes the constraint of <strong>"restrictiveness"</strong> on the concept of AND. In other words, it guarantees that the result of intersection is never more optimistic (larger) than any of its constituent components.
</p>

<h4 style="text-align: left;">
        b) Calculation and Comparison for $a = b = 0.9$ and the Reason for Preferring Algebraic Product
</h4>

<p style="text-align: left;">
        First, we calculate both operators for the given values:
</p>

<p style="text-align: center;">
        $$ t_{\min}(0.9, 0.9) = \min(0.9, 0.9) = 0.9 $$
        $$ t_{ap}(0.9, 0.9) = 0.9 \times 0.9 = 0.81 $$
</p>

<p style="text-align: left;">
        <strong>Comparison of results:</strong> As Theorem 3.2 predicts, the result of algebraic product ($0.81$) is less than or equal to the result of the minimum operator ($0.9$).
</p>

<p style="text-align: left;">
        <strong>Reason for preferring algebraic product (despite the simplicity of minimum):</strong>
</p>

<p style="text-align: left;">
        The problem statement explicitly mentions that these two signals are <strong>"independent."</strong> In probability theory and logic, the combination of two independent events is obtained through their multiplication (joint probability).
</p>

<p style="text-align: left;">
        The $\min(a,b)$ operator is a <i>non-interactive</i> operator, meaning that when $a=0.9$ and $b=0.9$, the output remains $0.9$ and the small uncertainty present in both signals (the remaining $0.1$ until complete certainty) is not aggregated. However, the algebraic product operator, by producing the value $0.81$, shows that since neither of the two signals individually has 100% certainty, their simultaneous combination carries more uncertainty, and the probability (or fuzzy degree) of their simultaneous occurrence decreases. This property, called <strong>interactivity</strong>, provides much more realistic and accurate behavior in modeling systems with independent variables (such as spam detection from different sources) compared to the $\min$ operator.
</p>

<h3 style="text-align: left;">
        Solution to Question 4.1 - De Morgan's Laws and Associated Class in Fuzzy Systems
</h3>

<h4 style="text-align: left;">
        Concept of Associated Class and De Morgan's Laws
</h4>

<p style="text-align: left;">
        In fuzzy set theory, for a system to remain logically consistent and integrated, the complement ($c$), intersection ($t$), and union ($s$) operators must be chosen to satisfy De Morgan's laws. These laws are expressed mathematically as follows:
</p>

<p style="text-align: center;">
        $$ c[s(a,b)] = t[c(a), c(b)] $$
        $$ c[t(a,b)] = s[c(a), c(b)] $$
</p>

<p style="text-align: left;">
        If a set of these three operators satisfies the above relationships, they are called <strong>"an associated class."</strong> Arbitrarily selecting operators from different families (e.g., Sugeno complement, Dombi union, and Yager intersection) without verifying this condition will almost certainly lead to violation of the above equalities.
</p>

<h4 style="text-align: left;">
        Logical Inconsistency in Fuzzy Inference Systems (FIS)
</h4>

<p style="text-align: left;">
        In designing the rule base of a fuzzy system, rules are typically formulated by an expert based on human language and logic (linguistic rules). The human mind intuitively and inherently considers classical logic equivalences, including De Morgan's equivalences, as valid. From the expert's perspective, the linguistic proposition <i>"if temperature is NOT high OR pressure is NOT high"</i> is exactly equivalent to the proposition <i>"if it is NOT the case that (temperature is high AND pressure is high)."</i>
</p>

<h4 style="text-align: left;">
        Consequences of Violating the Associated Class
</h4>

<p style="text-align: left;">
        If the designed operators do not form an associated class, the fuzzy system will compute two different numerical values (firing strengths) for these two completely equivalent propositions. Mathematically, the system calculations will result in a situation where:
</p>

<p style="text-align: center;">
        $$ c[t(\mu_A(x), \mu_B(x))] \neq s[c(\mu_A(x)), c(\mu_B(x))] $$
</p>

<h4 style="text-align: left;">
        Destructive Impact on System Performance
</h4>

<ul style="padding-left: 20px; text-align: left;">

<li style="margin-bottom: 12px;">
        <strong>Loss of logical robustness:</strong> If the rule base designer decides to change the grammatical expression of a rule to its logical equivalent, the control output (control surface) will change unexpectedly.
</li>

<li style="margin-bottom: 12px;">
        <strong>Unpredictable behavior:</strong> The fuzzy system will no longer accurately reflect the expert's intent and reasoning, because the underlying mathematical computations conflict with the conceptual logic governing the rules.
</li>

<li style="margin-bottom: 12px;">
        <strong>Difficulty in debugging and fine-tuning:</strong> Finding the root cause of control errors becomes very difficult in such a system, because the error is not due to parameters but due to a contradiction in the selected fuzzy algebra structure.
</li>

</ul>

<p style="text-align: left;">
        For this reason, it is always recommended to use operators that form a standard associated class (such as the Zadeh class consisting of minimum, maximum, and standard complement) when designing fuzzy systems.
</p>

<h3 style="text-align: left;">
        Solution to Question 2: Mackey-Glass Time Series Prediction Using ANFIS
</h3>

<p style="text-align: left;">
        The Mackey-Glass time series, modeled by a nonlinear delay differential equation, is one of the most important benchmarks for evaluating the prediction power of artificial intelligence models, particularly neuro-fuzzy systems (ANFIS). The dynamic formula of this system is as follows:
</p>

<p style="text-align: center;">
        $$ \frac{dx(t)}{dt} = \frac{a \cdot x(t-\tau)}{1+x^c(t-\tau)} - b \cdot x(t) $$
</p>

<h4 style="text-align: left;">
        System Dynamics Analysis
</h4>

<p style="text-align: left;">
        With parameter selection $a=0.2$, $b=0.1$, $c=10$, and most importantly the delay $\tau=17$, the system enters a <strong>chaotic regime</strong>. In this state, the system's phase space contains a strange attractor with fractal dimension that is highly sensitive to initial conditions. This characteristic makes long-term prediction using classical linear methods practically impossible, requiring universal approximators such as fuzzy systems.
</p>

<h4 style="text-align: left;">
        1. Data Preprocessing and Normalization
</h4>

<p style="text-align: left;">
        After generating 2000 samples through numerical solution of the equation (e.g., 4th-order Runge-Kutta method), the data must be standardized to prevent saturation of membership functions and accelerate the convergence of the learning algorithm in ANFIS. This is done by shifting the mean to zero and variance to one (Z-Score normalization):
</p>

<p style="text-align: center;">
        $$ \tilde{x}(t) = \frac{x(t) - \mu}{\sigma} $$
</p>

<h4 style="text-align: left;">
        2. Phase Space Reconstruction
</h4>

<p style="text-align: left;">
        According to the <strong>Takens' Embedding Theorem</strong>, to predict the future of a chaotic system using only one observable variable, we must consider a vector of past (delayed) values of the signal as network input. Assuming a delay of 3, the input vector $\mathbf{X}_t$ and desired output $y_t$ for training the ANFIS network are mapped as follows:
</p>

<p style="text-align: center;">
        $$ \mathbf{X}_t = \big[ \tilde{x}(t-3), \tilde{x}(t-2), \tilde{x}(t-1) \big]^T \quad \implies \quad y_t = \tilde{x}(t) $$
</p>

<h4 style="text-align: left;">
        3. ANFIS Architecture and Fuzzy Rule Generation
</h4>

<p style="text-align: left;">
        In the <code>genfis</code> function settings, when using the <strong>grid partitioning</strong> method, the input space is divided orthogonally. With 3 distinct inputs and assigning 2 membership functions (MFs) to each input, the total number of fuzzy rules inferred in the second layer of the network is calculated exponentially:
</p>

<p style="text-align: center;">
        $$ N_{rules} = \prod_{i=1}^{N_{inputs}} M_i = 2 \times 2 \times 2 = 2^3 = 8 \text{ Rules} $$
</p>

<h4 style="text-align: left;">
        4. Model Evaluation (Evaluation Metrics)
</h4>

<p style="text-align: left;">
        The dataset is split into ratios of 70% (training), 15% (validation), and 15% (testing). To accurately measure the performance of the fuzzy inference model, we use three statistical error metrics: Root Mean Square Error (RMSE), Mean Absolute Error (MAE), and Coefficient of Determination ($R^2$):
</p>

<p style="text-align: center;">
        $$ \text{RMSE} = \sqrt{\frac{1}{N} \sum_{i=1}^{N} (y_i - \hat{y}_i)^2} $$
        $$ \text{MAE} = \frac{1}{N} \sum_{i=1}^{N} |y_i - \hat{y}_i| $$
        $$ R^2 = 1 - \frac{\sum_{i=1}^{N} (y_i - \hat{y}_i)^2}{\sum_{i=1}^{N} (y_i - \bar{y})^2} $$
</p>

```python
# Libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from scipy.optimize import minimize
```

```python
# 1. Generate Mackey-Glass Time Series using RK4 Numerical Method
def mackey_glass_rk4(N=2000, tau=17, a=0.2, b=0.1, c=10, dt=1.0):
    """
    Generate Mackey-Glass time series data using 4th Order Runge-Kutta method
    
    Parameters:
    N: Number of samples to generate
    tau: Time delay parameter (standard = 17)
    a, b, c: Equation parameters (standard values: a=0.2, b=0.1, c=10)
    dt: Time step size
    
    Returns:
    x: Generated time series data
    """
    history_len = int(tau / dt)
    time_steps = N + history_len + 500  # Add buffer to remove transient effects
    x = np.zeros(time_steps)
    x[:history_len] = 1.2  # Standard initial conditions
    for t in range(history_len, time_steps - 1):
        x_tau = x[t - int(tau/dt)]
        
        # Calculate RK4 coefficients for higher accuracy in solving differential equation
        # The equation: dx/dt = (a * x(t-tau)) / (1 + x(t-tau)^c) - b * x(t)
        k1 = dt * ((a * x_tau) / (1 + x_tau**c) - b * x[t])
        k2 = dt * ((a * x_tau) / (1 + x_tau**c) - b * (x[t] + 0.5 * k1))
        k3 = dt * ((a * x_tau) / (1 + x_tau**c) - b * (x[t] + 0.5 * k2))
        k4 = dt * ((a * x_tau) / (1 + x_tau**c) - b * (x[t] + k3))
        
        x[t+1] = x[t] + (k1 + 2*k2 + 2*k3 + k4) / 6.0
        
    # Return exactly the last N samples (removing initial transient effects)
    return x[-N:]

# Generate 2000 data points
data = mackey_glass_rk4(N=2000)
```

```python
# 2. Normalize Data Distribution (Mean=0, Std=1)
mu = np.mean(data)
sigma = np.std(data)
data_norm = (data - mu) / sigma
```

```python
# 3. Phase Space Reconstruction (Creating Input-Output Pairs with Delay=3)
delay = 3
X_features = []
Y_targets = []

# Create input-output pairs
for i in range(delay, len(data_norm)):
    # Input: [x(t-3), x(t-2), x(t-1)]
    X_features.append([data_norm[i-3], data_norm[i-2], data_norm[i-1]])
    # Output: x(t)
    Y_targets.append(data_norm[i])

X_features = np.array(X_features)
Y_targets = np.array(Y_targets)
```

```python
# 4. Split Dataset (70% Training, 15% Validation, 15% Testing)
total_samples = len(X_features)
train_end = int(0.70 * total_samples)
val_end = int(0.85 * total_samples)

X_train, Y_train = X_features[:train_end], Y_targets[:train_end]
X_val, Y_val     = X_features[train_end:val_end], Y_targets[train_end:val_end]
X_test, Y_test   = X_features[val_end:], Y_targets[val_end:]

# Print information for verification
print(f"Total structured samples: {total_samples}")
print(f"Train Set Shape: X={X_train.shape}, Y={Y_train.shape}")
print(f"Val Set Shape:   X={X_val.shape}, Y={Y_val.shape}")
print(f"Test Set Shape:  X={X_test.shape}, Y={Y_test.shape}")
```

```python
# 5. Define ANFIS (Adaptive Neuro-Fuzzy Inference System) Class
class ANFIS:
    """
    Adaptive Neuro-Fuzzy Inference System for time series prediction
    
    This implementation uses:
    - Gaussian membership functions
    - Product inference engine
    - Weighted average defuzzification
    - Hybrid learning (backpropagation + least squares)
    """
    
    def __init__(self, n_inputs=3, n_mf=3):
        """
        Initialize ANFIS with specified structure
        
        Parameters:
        n_inputs: Number of input features
        n_mf: Number of membership functions per input
        """
        self.n_inputs = n_inputs
        self.n_mf = n_mf
        
        # Initialize membership function parameters (mean and sigma for Gaussian)
        # Each input has n_mf Gaussian functions: [mean1, sigma1, mean2, sigma2, ...]
        self.mf_params = np.random.uniform(-1, 1, (n_inputs, 2 * n_mf))
        
        # Initialize consequent parameters (linear coefficients for each rule)
        # Each rule has: [p, q, r, s] for output: y = p*x1 + q*x2 + r*x3 + s
        self.consequent = np.random.randn(n_mf ** n_inputs, n_inputs + 1)
        
        self.n_rules = n_mf ** n_inputs
        self.input_ranges = None
        
    def gaussian_mf(self, x, mean, sigma):
        """
        Gaussian membership function: exp(-(x-mean)^2 / (2*sigma^2))
        """
        return np.exp(-((x - mean) ** 2) / (2 * sigma ** 2 + 1e-10))
    
    def get_mf_indices(self, rule_idx):
        """
        Convert rule index to combination of membership function indices
        Example: for n_inputs=3, n_mf=3, rule 5 -> [1, 2, 0]
        """
        indices = []
        temp = rule_idx
        for i in range(self.n_inputs):
            indices.append(temp % self.n_mf)
            temp = temp // self.n_mf
        return indices
    
    def forward(self, X):
        """
        Forward pass: compute output for given inputs
        
        Parameters:
        X: Input data of shape (n_samples, n_inputs)
        
        Returns:
        Output predictions
        """
        n_samples = X.shape[0]
        output = np.zeros(n_samples)
        
        # For each input sample
        for s in range(n_samples):
            # Compute membership degrees for all inputs
            mf_activations = []
            for i in range(self.n_inputs):
                mf_values = []
                for j in range(self.n_mf):
                    mean = self.mf_params[i, 2*j]
                    sigma = np.abs(self.mf_params[i, 2*j + 1]) + 0.01
                    mf_values.append(self.gaussian_mf(X[s, i], mean, sigma))
                mf_activations.append(mf_values)
            
            # Compute firing strength for each rule (product of input MFs)
            firing_strengths = np.zeros(self.n_rules)
            for rule_idx in range(self.n_rules):
                mf_indices = self.get_mf_indices(rule_idx)
                strength = 1.0
                for i in range(self.n_inputs):
                    strength *= mf_activations[i][mf_indices[i]]
                firing_strengths[rule_idx] = strength
            
            # Normalize firing strengths
            total_firing = np.sum(firing_strengths) + 1e-10
            normalized = firing_strengths / total_firing
            
            # Compute weighted output
            y_pred = 0.0
            for rule_idx in range(self.n_rules):
                # Linear consequent: y = p*x1 + q*x2 + r*x3 + s
                rule_output = np.dot(self.consequent[rule_idx], 
                                   np.append(X[s], 1.0))
                y_pred += normalized[rule_idx] * rule_output
            
            output[s] = y_pred
            
        return output
    
    def predict(self, X):
        """
        Predict output for given inputs
        """
        return self.forward(X)
    
    def train(self, X_train, Y_train, X_val, Y_val, epochs=50, lr=0.01):
        """
        Train ANFIS using hybrid learning
        
        Parameters:
        X_train, Y_train: Training data
        X_val, Y_val: Validation data (for early stopping)
        epochs: Number of training epochs
        lr: Learning rate for backpropagation
        """
        n_samples = X_train.shape[0]
        best_val_rmse = float('inf')
        patience = 10  # Early stopping patience
        patience_counter = 0
        
        # Store training history
        train_rmse_history = []
        val_rmse_history = []
        
        for epoch in range(epochs):
            # Shuffle training data
            indices = np.random.permutation(n_samples)
            X_shuffled = X_train[indices]
            Y_shuffled = Y_train[indices]
            
            total_loss = 0.0
            
            # Batch training (gradient descent)
            for s in range(n_samples):
                x_sample = X_shuffled[s]
                y_target = Y_shuffled[s]
                
                # --- Forward pass ---
                # Compute membership values
                mf_values = []
                mf_gradients = []  # Store gradients for MF parameters
                
                for i in range(self.n_inputs):
                    mf_row = []
                    grad_row = []
                    for j in range(self.n_mf):
                        mean = self.mf_params[i, 2*j]
                        sigma = np.abs(self.mf_params[i, 2*j + 1]) + 0.01
                        mf_val = self.gaussian_mf(x_sample[i], mean, sigma)
                        mf_row.append(mf_val)
                        
                        # Store gradient components for mean and sigma
                        grad_mean = mf_val * (x_sample[i] - mean) / (sigma**2)
                        grad_sigma = mf_val * ((x_sample[i] - mean)**2) / (sigma**3)
                        grad_row.append((grad_mean, grad_sigma))
                    mf_values.append(mf_row)
                    mf_gradients.append(grad_row)
                
                # Compute rule firing strengths
                firing_strengths = np.zeros(self.n_rules)
                for rule_idx in range(self.n_rules):
                    mf_indices = self.get_mf_indices(rule_idx)
                    strength = 1.0
                    for i in range(self.n_inputs):
                        strength *= mf_values[i][mf_indices[i]]
                    firing_strengths[rule_idx] = strength
                
                total_firing = np.sum(firing_strengths) + 1e-10
                normalized = firing_strengths / total_firing
                
                # Compute output and error
                y_pred = 0.0
                for rule_idx in range(self.n_rules):
                    rule_output = np.dot(self.consequent[rule_idx], 
                                       np.append(x_sample, 1.0))
                    y_pred += normalized[rule_idx] * rule_output
                
                error = y_pred - y_target
                total_loss += error**2
                
                # --- Backward pass: Update consequent parameters (Least Squares) ---
                # Simplified: Use gradient descent for consequent
                for rule_idx in range(self.n_rules):
                    weight = normalized[rule_idx]
                    inputs_with_bias = np.append(x_sample, 1.0)
                    # Gradient: d(error)/d(p) = weight * error * x
                    self.consequent[rule_idx] -= lr * 2 * error * weight * inputs_with_bias
                
                # --- Backward pass: Update MF parameters (Backpropagation) ---
                for i in range(self.n_inputs):
                    for j in range(self.n_mf):
                        # Compute gradient for mean and sigma
                        grad_mean_sum = 0.0
                        grad_sigma_sum = 0.0
                        
                        # Contribution from all rules involving this MF
                        for rule_idx in range(self.n_rules):
                            mf_indices = self.get_mf_indices(rule_idx)
                            if mf_indices[i] == j:
                                # Compute derivative of normalized firing strength
                                weight = normalized[rule_idx]
                                strength = firing_strengths[rule_idx]
                                d_weight = (1.0/total_firing - weight/total_firing)
                                
                                # Derivative of MF with respect to parameters
                                d_mf_mean = mf_gradients[i][j][0]
                                d_mf_sigma = mf_gradients[i][j][1]
                                
                                # Contribution to error gradient
                                rule_output = np.dot(self.consequent[rule_idx], 
                                                   np.append(x_sample, 1.0))
                                grad_mean_sum += 2 * error * d_weight * rule_output * d_mf_mean
                                grad_sigma_sum += 2 * error * d_weight * rule_output * d_mf_sigma
                        
                        # Update mean and sigma
                        self.mf_params[i, 2*j] -= lr * grad_mean_sum
                        self.mf_params[i, 2*j + 1] -= lr * grad_sigma_sum
                        # Ensure sigma stays positive
                        if self.mf_params[i, 2*j + 1] < 0.01:
                            self.mf_params[i, 2*j + 1] = 0.01
            
            # Calculate RMSE for training and validation
            train_pred = self.forward(X_train)
            val_pred = self.forward(X_val)
            train_rmse = np.sqrt(mean_squared_error(Y_train, train_pred))
            val_rmse = np.sqrt(mean_squared_error(Y_val, val_pred))
            
            train_rmse_history.append(train_rmse)
            val_rmse_history.append(val_rmse)
            
            if epoch % 10 == 0:
                print(f"Epoch {epoch}: Train RMSE = {train_rmse:.4f}, Val RMSE = {val_rmse:.4f}")
            
            # Early stopping
            if val_rmse < best_val_rmse:
                best_val_rmse = val_rmse
                patience_counter = 0
                # Save best parameters
                best_mf_params = self.mf_params.copy()
                best_consequent = self.consequent.copy()
            else:
                patience_counter += 1
                if patience_counter >= patience:
                    print(f"Early stopping at epoch {epoch}")
                    self.mf_params = best_mf_params
                    self.consequent = best_consequent
                    break
        
        return train_rmse_history, val_rmse_history
```

```python
# 6. Define Model Evaluation Function
def evaluate_anfis_performance(y_true, y_pred, dataset_name="Dataset"):
    """
    Calculate and print model evaluation metrics
    """
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mae = mean_absolute_error(y_true, y_pred)
    r2 = r2_score(y_true, y_pred)
    
    print("-" * 40)
    print(f"{dataset_name} Performance Metrics:")
    print(f"RMSE : {rmse:.6f}")
    print(f"MAE  : {mae:.6f}")
    print(f"R^2  : {r2:.6f}")
    print("-" * 40)
    
    return rmse, mae, r2
```

<h3 style="text-align: left;">
        Solution to Question 3: Designing a Fuzzy System Using the Wang-Mendel Method Based on Lookup Table
</h3>

<p style="text-align: left;">
        The Wang-Mendel method is a systematic and efficient approach for extracting fuzzy rules from a set of numerical input-output data. The goal of this method is to construct a rule base for approximating nonlinear functions. The algorithm consists of 5 basic steps, which are described below for the given function:
</p>

<h4 style="text-align: left;">
        Step 1: Partitioning the Input and Output Spaces into Fuzzy Regions
</h4>

<p style="text-align: left;">
        According to the problem statement, the input variables $x_1$ and $x_2$ are in the range $[-1, 1]$. Given the relationship $y = \sin(\pi x_1) + 0.5 x_2^2$, the output range $y$ will be in the interval $[-1, 1.5]$.
</p>

<p style="text-align: left;">
        For each variable, we define 7 fuzzy sets with triangular membership functions named as follows:
</p>

<p style="text-align: center;">
        $$ \{NB, NM, NS, ZE, PS, PM, PB\} $$
</p>

<h4 style="text-align: left;">
        Step 2: Extracting Fuzzy Rules from Training Data
</h4>

<p style="text-align: left;">
        For each training sample $(x_{1}^{(i)}, x_{2}^{(i)}, y^{(i)})$, we first calculate its membership degree in all fuzzy sets. Then, for each variable, the set with the highest membership degree is selected as the dominant label.
</p>

<p style="text-align: left;">
        Suppose the dominant labels for a sample are $A^*$, $B^*$, and $C^*$, respectively. The corresponding rule is extracted as follows:
</p>

<p style="text-align: center;">
        $$ \text{IF } x_1 \text{ is } A^* \text{ AND } x_2 \text{ is } B^* \text{ THEN } y \text{ is } C^* $$
</p>

<h4 style="text-align: left;">
        Step 3: Assigning a Degree to Each Rule
</h4>

<p style="text-align: left;">
        Since noisy or conflicting data may produce rules with identical antecedents but different consequents, we must assign a degree to each rule to resolve conflicts in the Lookup Table. The degree of the rule extracted from the $i$-th sample is:
</p>

<p style="text-align: center;">
        $$ Degree^{(i)} = \mu_{A^*}(x_{1}^{(i)}) \times \mu_{B^*}(x_{2}^{(i)}) \times \mu_{C^*}(y^{(i)}) $$
</p>

<h4 style="text-align: left;">
        Steps 4 and 5: Forming the Rule Base and Inference System
</h4>

<p style="text-align: left;">
        In case of conflicts (rules with the same "IF" part but different "THEN" part), the rule with the <strong>maximum degree</strong> is retained in the rule base (Lookup Table) and the rest are eliminated.
</p>

<p style="text-align: left;">
        Finally, using the inference engine (typically product of operators) and the center of average defuzzifier, the final output of the fuzzy system for function approximation is calculated as:
</p>

<p style="text-align: center;">
        $$ f(x) = \frac{\sum_{l=1}^{M} \bar{y}^l \cdot \left( \mu_{A^l}(x_1) \times \mu_{B^l}(x_2) \right)}{\sum_{l=1}^{M} \left( \mu_{A^l}(x_1) \times \mu_{B^l}(x_2) \right)} $$
</p>

<p style="text-align: left;">
        where $M$ is the total number of valid rules in the Lookup Table and $\bar{y}^l$ is the centroid of the output fuzzy set in the $l$-th rule.
</p>

</div>

```python
# 1. Data Generation
np.random.seed(42)
N = 500
x1 = np.random.uniform(-1, 1, N)
x2 = np.random.uniform(-1, 1, N)
y = np.sin(np.pi * x1) + 0.5 * (x2**2)
```

```python
# 2. Define Fuzzy Membership Functions
labels = ['NB', 'NM', 'NS', 'ZE', 'PS', 'PM', 'PB']
num_labels = len(labels)

# Calculate min/max for domains
x1_min, x1_max = -1.0, 1.0
x2_min, x2_max = -1.0, 1.0
y_min, y_max = -1.0, 1.5 # Based on function bounds

def create_centers(min_val, max_val, n):
    return np.linspace(min_val, max_val, n)

centers_x1 = create_centers(x1_min, x1_max, num_labels)
centers_x2 = create_centers(x2_min, x2_max, num_labels)
centers_y = create_centers(y_min, y_max, num_labels)

def triangular_mf(x, centers, idx):
    """Triangular membership function evaluator."""
    c = centers[idx]
    width = (centers[-1] - centers[0]) / (len(centers) - 1)
    
    # Left and right bounds for the triangle
    left = c - width
    right = c + width
    
    # Membership calculation
    if x <= left or x >= right:
        return 0.0
    elif x == c:
        return 1.0
    elif x < c:
        return (x - left) / (c - left)
    else:
        return (right - x) / (right - c)

def get_dominant_fuzzy_label(val, centers):
    """Finds the label with max membership degree."""
    memberships = [triangular_mf(val, centers, i) for i in range(len(centers))]
    max_idx = np.argmax(memberships)
    max_val = memberships[max_idx]
    return max_idx, max_val
```

```python
# 3. Rule Extraction (Wang-Mendel Method)
lookup_table = {} # Dictionary to store Rule Base: {(x1_idx, x2_idx): (y_idx, degree)}

for i in range(N):
    # Find dominant sets and their degrees
    idx_x1, mu_x1 = get_dominant_fuzzy_label(x1[i], centers_x1)
    idx_x2, mu_x2 = get_dominant_fuzzy_label(x2[i], centers_x2)
    idx_y, mu_y = get_dominant_fuzzy_label(y[i], centers_y)
    
    # Calculate Rule Degree
    rule_degree = mu_x1 * mu_x2 * mu_y
    
    # IF part of the rule (antecedent)
    rule_if = (idx_x1, idx_x2)
    
    # Resolve conflicts in Lookup Table
    if rule_if in lookup_table:
        # Conflict: Keep the rule with the maximum degree
        if rule_degree > lookup_table[rule_if][1]:
            lookup_table[rule_if] = (idx_y, rule_degree)
    else:
        # No conflict: add new rule
        lookup_table[rule_if] = (idx_y, rule_degree)
```

```python
# 4. Display Extracted Rules
print(f"Total rules extracted in Lookup Table: {len(lookup_table)} / {num_labels*num_labels}")
print("\nSample Rules from the Lookup Table:")
print("-" * 50)
count = 0
for (x1_idx, x2_idx), (y_idx, degree) in lookup_table.items():
    print(f"IF x1 is {labels[x1_idx]:2s} AND x2 is {labels[x2_idx]:2s} THEN y is {labels[y_idx]:2s} (Degree: {degree:.4f})")
    count += 1
    if count >= 10: # Print only first 10 rules for brevity
        print("... (and more)")
        break
```

```python
labels = ['NB', 'NM', 'NS', 'ZE', 'PS', 'PM', 'PB']
x_domain = np.linspace(-1, 1, 400)
y_domain = np.linspace(-1, 1.5, 400)

centers_x = np.linspace(-1, 1, 7)
centers_y = np.linspace(-1, 1.5, 7)

def triangular_mf_array(x_array, centers, idx):
    c = centers[idx]
    width = (centers[-1] - centers[0]) / (len(centers) - 1)
    left, right = c - width, c + width
    y_mf = np.zeros_like(x_array)
    for i, x in enumerate(x_array):
        if left < x <= c:
            y_mf[i] = (x - left) / (c - left)
        elif c < x < right:
            y_mf[i] = (right - x) / (right - c)
    return y_mf
fig, axs = plt.subplots(1, 2, figsize=(12, 4))
for i in range(7):
    axs[0].plot(x_domain, triangular_mf_array(x_domain, centers_x, i), label=labels[i])
axs[0].set_title("Membership Functions for x1 and x2")
axs[0].legend()
axs[0].grid(True)

for i in range(7):
    axs[1].plot(y_domain, triangular_mf_array(y_domain, centers_y, i), label=labels[i])
axs[1].set_title("Membership Functions for y")
axs[1].legend()
axs[1].grid(True)

plt.tight_layout()
plt.show()
```

<div dir="ltr" style="font-family: 'Segoe UI', sans-serif; line-height: 1.8; text-align: left;">

<h4 style="text-align: left;">
        Part 2: Rule Extraction and Rule Degree Calculation
</h4>

<p style="text-align: left;">
        In this section, for each of the $N$ data points in the training dataset of the form $(x_1^{(i)}, x_2^{(i)}, y^{(i)})$, we perform the following steps:
</p>

<p style="text-align: left;">
        1. First, we determine the fuzzy set with the highest membership degree (dominant set) for each of the input and output variables. In other words, fuzzy sets $A^{(i)}$ for $x_1$, $B^{(i)}$ for $x_2$, and $C^{(i)}$ for $y$ are selected such that the membership degree of the data point in them is maximum.
</p>

<p style="text-align: left;">
        2. After determining the corresponding fuzzy sets, we calculate the rule degree (degree of the rule) using the product T-norm as follows:
</p>

<p style="text-align: center;">
        $$ Degree^{(i)} = \mu_{A^{(i)}}(x_1^{(i)}) \times \mu_{B^{(i)}}(x_2^{(i)}) \times \mu_{C^{(i)}}(y^{(i)}) $$
</p>

<p style="text-align: left;">
        At this stage, all candidate rules along with their degrees are stored for processing and conflict resolution in the next step.
</p>

</div>

```python
# Part 2: Extracting candidate rules and calculating their degrees

candidate_rules = []

for i in range(N):
    # Get dominant fuzzy label index and membership degree for x1, x2, and y
    idx_x1, mu_x1 = get_dominant_fuzzy_label(x1[i], centers_x1)
    idx_x2, mu_x2 = get_dominant_fuzzy_label(x2[i], centers_x2)
    idx_y, mu_y = get_dominant_fuzzy_label(y[i], centers_y)
    
    # Calculate rule degree using product t-norm
    rule_degree = mu_x1 * mu_x2 * mu_y
    
    # Save candidate rule format: ((idx_x1, idx_x2), idx_y, degree)
    candidate_rules.append({
        'antecedent': (idx_x1, idx_x2),
        'consequent': idx_y,
        'degree': rule_degree
    })

# Print information about extracted candidate rules
print(f"Successfully generated {len(candidate_rules)} candidate rules from {N} data points.")
```

<div dir="ltr" style="font-family: 'Segoe UI', sans-serif; line-height: 1.8; text-align: left;">

<h4 style="text-align: left;">
        Part 3: Conflict Resolution Between Rules
</h4>

<p style="text-align: left;">
        A conflict occurs when two or more candidate rules have the same antecedent (IF part) but different consequents (THEN part), meaning that similar inputs are mapped to different outputs. This phenomenon typically arises due to noise in the data or the nonlinear nature of the system.
</p>

<p style="text-align: left;">
        To resolve this conflict in the Wang-Mendel method, the following strategy is employed:
</p>

<p style="text-align: left;">
        Among all rules that share the same antecedent, only the rule with the <strong>maximum rule degree</strong> is retained, and the remaining conflicting rules are eliminated. This approach ensures that the strongest and most reliable relationship between input and output remains in the rule base.
</p>

</div>

```python
# Part 3: Resolving conflicts based on the maximum rule degree

resolved_rules = {}

for rule in candidate_rules:
    antecedent = rule['antecedent']
    consequent = rule['consequent']
    degree = rule['degree']
    
    if antecedent in resolved_rules:
        # If conflict exists, compare degrees and keep the one with max degree
        if degree > resolved_rules[antecedent]['degree']:
            resolved_rules[antecedent] = {
                'consequent': consequent,
                'degree': degree
            }
    else:
        # If no conflict yet, add the rule to the resolved rules dictionary
        resolved_rules[antecedent] = {
            'consequent': consequent,
            'degree': degree
        }

print(f"Number of rules after resolving conflicts: {len(resolved_rules)}")
```

<div dir="ltr" style="font-family: 'Segoe UI', sans-serif; line-height: 1.8; text-align: left;">

<h4 style="text-align: left;">
        Part 4: Forming the Rule Base (Lookup Table)
</h4>

<p style="text-align: left;">
        In this section, we store the final rules obtained from the conflict resolution stage in the structure of the fuzzy system rule base, which is organized as a lookup table.
</p>

<p style="text-align: left;">
        Since we have defined 7 membership functions for each of the input variables $x_1$ and $x_2$, the rule base table will have a maximum dimension of $7 \times 7 = 49$ cells. Whether the cells of this table are filled or empty depends on the distribution of the training data.
</p>

</div>

```python
# Part 4: Populate the final lookup table and display the rule base

# lookup_table is structured as: {(idx_x1, idx_x2): (idx_y, degree)}
lookup_table = {}

for antecedent, info in resolved_rules.items():
    lookup_table[antecedent] = (info['consequent'], info['degree'])

# Print final rule base details
print(f"Final Lookup Table contains {len(lookup_table)} active rules out of 49 possible spaces.\n")
print("Sample of Final Extracted Rules:")
print("-" * 60)
count = 0
for (idx_x1, idx_x2), (idx_y, degree) in lookup_table.items():
    print(f"IF x1 is {labels[idx_x1]:2s} AND x2 is {labels[idx_x2]:2s} THEN y is {labels[idx_y]:2s} | Degree: {degree:.4f}")
    count += 1
    if count >= 10:
        print("... (and more rules)")
        break
```

<div dir="ltr" style="font-family: 'Segoe UI', sans-serif; line-height: 1.8; text-align: left;">

<h4 style="text-align: left;">
        Part 5: Fuzzy Inference
</h4>

<p style="text-align: left;">
        For each new input data point of the form $(x_1^{test}, x_2^{test})$, the fuzzy inference engine performs the following steps to determine the final crisp output value:
</p>

<p style="text-align: left;">
        <strong>1. Fuzzification and Calculation of Rule Firing Strength:</strong>
</p>

<p style="text-align: left;">
        For each rule in the rule base (Lookup Table), the degree of matching of the inputs with the rule antecedent is calculated using the product T-norm:
</p>

<p style="text-align: center;">
        $$ w_l = \mu_{A^l}(x_1^{test}) \times \mu_{B^l}(x_2^{test}) $$
</p>

<p style="text-align: left;">
        <strong>2. Centroid Defuzzification:</strong>
</p>

<p style="text-align: left;">
        Since the output membership functions are symmetric, the output centroid can be approximated as the weighted average of the centroids of the output fuzzy sets of the activated rules ($\bar{y}^l$):
</p>

<p style="text-align: center;">
        $$ y_{pred} = \frac{\sum_{l=1}^{M} w_l \cdot \bar{y}^l}{\sum_{l=1}^{M} w_l} $$
</p>

<p style="text-align: left;">
        where $M$ is the total number of active rules in the rule table.
</p>

</div>

```python
# Part 5: Defining the Fuzzy Inference engine function

def fuzzy_inference(test_x1, test_x2, rule_base, cx1, cx2, cy):
    """
    Computes the defuzzified output for a given test input using the Centroid method.
    """
    numerator = 0.0
    denominator = 0.0
    
    # Iterate over all rules stored in the lookup table
    for (idx_x1, idx_x2), (idx_y, _) in rule_base.items():
        # Fuzzify inputs for the current rule antecedent
        mu_x1 = triangular_mf(test_x1, cx1, idx_x1)
        mu_x2 = triangular_mf(test_x2, cx2, idx_x2)
        
        # Calculate firing strength of the rule using product t-norm
        w_l = mu_x1 * mu_x2
        
        # Get the center of the consequent fuzzy set
        y_center = cy[idx_y]
        
        # Aggregate the weighted output
        numerator += w_l * y_center
        denominator += w_l
        
    # Handle the case where no rules are fired to prevent division by zero
    if denominator == 0.0:
        return 0.0
        
    return numerator / denominator
```

<h4 style="text-align: left;">
        1. Why is it necessary to eliminate rule conflicts in the Wang-Mendel method?
</h4>

<p style="text-align: left;">
        In this method, rules are directly extracted from raw numerical samples. Since data may contain noise or the system's dynamic behavior may be dense and nonlinear in some regions, it is entirely natural that data points with very close input values (which fall into the same category or fuzzy antecedent) produce different output values. This leads to the creation of multiple rules with the same antecedent but different consequents (conflicts). Since the fuzzy system needs a consistent rule base to provide stable and definite responses, only one specific output must be determined for each input region. The conflict resolution process, by selecting the rule with the highest overall membership degree (degree of validity), filters out noise and preserves the best approximation for each cell of the space.
</p>

<h4 style="text-align: left;">
        2. What is the effect of increasing the number of membership functions on the number of rules and system accuracy?
</h4>

<ul style="padding-left: 20px; text-align: left;">

<li style="margin-bottom: 12px;">
        <strong>Number of rules:</strong> The potential number of rules in fuzzy systems has an exponential relationship with the number of input membership functions. If $n$ is the number of input variables and $M$ is the number of membership functions for each variable, the maximum number of rules will be $M^n$. Increasing $M$ drastically increases the number of rules and leads to the "curse of dimensionality" challenge.
</li>

<li style="margin-bottom: 12px;">
        <strong>System accuracy:</strong> Initially, increasing the number of membership functions creates a finer-grained partitioning of the state space, allowing the model to approximate complex nonlinear behaviors with more detail and lower error (reduction in RMSE on training data). However, if the number of membership functions exceeds the capacity of the training data, many regions of the space will remain without rules or weak rules will be created. This phenomenon leads to overfitting and reduces the system's generalization ability on new test data.
</li>

</ul>

<h4 style="text-align: left;">
        3. What is the most important advantage of using a Lookup Table in fuzzy systems?
</h4>

<p style="text-align: left;">
        The most important advantage of using a Lookup Table is <strong>"very high speed in one-pass learning and its structural simplicity."</strong> Unlike optimization methods and iterative algorithms (such as ANFIS or fuzzy systems based on genetic algorithms) that require heavy gradient computations, multiple iterations (epochs), and long convergence times, the Lookup Table method builds the entire rule base with just a single pass through the training data. This characteristic makes this method highly efficient and economical for real-time systems, big data, and environments with limited computational power.
</p>

</div>

This approach achieves both high accuracy and complete engineering interpretability.

<h3 style="text-align: left; margin-bottom: 10px;">Step 1: 5-Layer ANFIS Architecture and Mathematical Formulation</h3>

<p style="text-align: left;">
        The proposed ANFIS system in this project has two inputs ($x_1 = V$ and $x_2 = T$) and for each input, 2 Gaussian membership functions are considered. Therefore, the rule base will consist of $2 \times 2 = 4$ fuzzy rules. The architecture of this system consists of 5 feedforward layers as described below:
</p>

<ul style="padding-left: 20px; text-align: left;">

<li style="margin-bottom: 15px;">
        <strong>Layer 1 (Fuzzification Layer):</strong> In this layer, the membership degree of each input to the fuzzy sets is calculated. The parameters of this layer $(c, \sigma)$ are called <strong>premise parameters</strong>.
        <br>
        $$ O_{1,i} = \mu_{A_i}(x) = \exp\left(-\frac{(x - c_i)^2}{2\sigma_i^2}\right) $$
</li>

<li style="margin-bottom: 15px;">
        <strong>Layer 2 (Product or T-norm Layer):</strong> This layer calculates the firing strength of each rule by multiplying the membership degrees of the inputs. If rule $i$ has antecedents $A_i$ for $V$ and $B_i$ for $T$:
        <br>
        $$ O_{2,i} = w_i = \mu_{A_i}(V) \times \mu_{B_i}(T) $$
</li>

<li style="margin-bottom: 15px;">
        <strong>Layer 3 (Normalization Layer):</strong> The firing strength of each rule is normalized relative to the sum of the strengths of all rules.
        <br>
        $$ O_{3,i} = \bar{w}_i = \frac{w_i}{\sum_{j=1}^{4} w_j} $$
</li>

<li style="margin-bottom: 15px;">
        <strong>Layer 4 (TSK Linear Consequent Layer):</strong> Here, the normalized weight is multiplied by the linear function output of the same rule. The parameters $(p_i, q_i, r_i)$ are <strong>consequent parameters</strong>.
        <br>
        $$ O_{4,i} = \bar{w}_i f_i = \bar{w}_i (p_i V + q_i T + r_i) $$
</li>

<li style="margin-bottom: 0px;">
        <strong>Layer 5 (Total Output Layer):</strong> This layer calculates the sum of all signals from the previous layer as the final system output (fuel consumption prediction).
        <br>
        $$ O_{5,i} = y_{pred} = \sum_{i=1}^{4} \bar{w}_i f_i = \frac{\sum_{i=1}^{4} w_i f_i}{\sum_{i=1}^{4} w_i} $$
</li>

</ul>

</div>

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# Step 1 Addition: ANFIS Architecture Diagram
fig, ax = plt.subplots(figsize=(12, 6))
ax.axis('off')

# Define layer positions and nodes
layers = {
    'Input': {'x': 0, 'y': [2, 4], 'labels': ['V', 'T']},
    'Layer 1\n(Fuzzification)': {'x': 2, 'y': [1, 3, 5, 7], 'labels': ['A1', 'A2', 'B1', 'B2'], 'shape': 'square'},
    'Layer 2\n(Product / П)': {'x': 4, 'y': [1.5, 3.5, 5.5, 7.5], 'labels': ['w1', 'w2', 'w3', 'w4'], 'shape': 'circle'},
    'Layer 3\n(Normalization / N)': {'x': 6, 'y': [1.5, 3.5, 5.5, 7.5], 'labels': ['w1_bar', 'w2_bar', 'w3_bar', 'w4_bar'], 'shape': 'circle'},
    'Layer 4\n(Consequent)': {'x': 8, 'y': [1.5, 3.5, 5.5, 7.5], 'labels': ['w1*f1', 'w2*f2', 'w3*f3', 'w4*f4'], 'shape': 'square'},
    'Layer 5\n(Output / Σ)': {'x': 10, 'y': [4.5], 'labels': ['Σ'], 'shape': 'circle'},
    'Output': {'x': 12, 'y': [4.5], 'labels': ['y_pred']}
}

# Draw nodes and lines (conceptual representation)
for layer_name, data in layers.items():
    x = data['x']
    plt.text(x, 8.5, layer_name, ha='center', va='center', fontweight='bold', fontsize=10)
    for i, y in enumerate(data['y']):
        if 'shape' in data:
            if data['shape'] == 'square': # Adaptive nodes
                rect = patches.Rectangle((x-0.4, y-0.4), 0.8, 0.8, fill=True, color='lightblue', ec='black')
                ax.add_patch(rect)
            elif data['shape'] == 'circle': # Fixed nodes
                circle = patches.Circle((x, y), 0.4, fill=True, color='lightgreen', ec='black')
                ax.add_patch(circle)
        
        # Add labels
        plt.text(x, y, data['labels'][i], ha='center', va='center', fontsize=9)

# Draw connecting lines (simplified for clarity)
# Input to Layer 1
ax.plot([0.5, 1.6], [2, 1], 'k-', alpha=0.5)
ax.plot([0.5, 1.6], [2, 3], 'k-', alpha=0.5)
ax.plot([0.5, 1.6], [4, 5], 'k-', alpha=0.5)
ax.plot([0.5, 1.6], [4, 7], 'k-', alpha=0.5)

# Layer 4 to Layer 5
for y in layers['Layer 4\n(Consequent)']['y']:
    ax.plot([8.5, 9.6], [y, 4.5], 'k-', alpha=0.5)

# Layer 5 to Output
ax.plot([10.5, 11.5], [4.5, 4.5], 'k-', alpha=0.5)

plt.title("Step 1: ANFIS 5-Layer Architecture Diagram", fontsize=14, pad=20)
plt.xlim(-1, 13)
plt.ylim(0, 9)
plt.show()
```

<h3 style="text-align: left; margin-bottom: 10px;">Step 2: Hybrid Learning Algorithm</h3>

<p style="text-align: left;">
        ANFIS training using the Jang (J.S.R. Jang) method employs a two-stage strategy in each epoch to avoid getting stuck in local minima and significantly increase convergence speed.
</p>

<ul style="padding-left: 20px; text-align: left;">

<li style="margin-bottom: 15px;">
        <strong>Forward Pass:</strong> In this stage, the Gaussian parameters (centers and widths) are assumed fixed. Data propagates up to layer 3 and the values of $\bar{w}_i$ are calculated. The total system output can be written as a linear combination of <strong>consequent parameters</strong>:
        <br>
        $$ y = (\bar{w}_1 V)p_1 + (\bar{w}_1 T)q_1 + (\bar{w}_1)r_1 + \dots + (\bar{w}_4)r_4 $$
        This equation can be written in matrix form as $A\theta = y$, where $A$ is the design matrix (consisting of the combination of inputs and normalized weights) and $\theta$ is the vector of consequent parameters. To find the optimal $\theta$ in one step, we use the <strong>Least Squares Error (LSE)</strong> method:
        <br>
        $$ \theta^* = (A^T A)^{-1} A^T y_{true} $$
</li>

<li style="margin-bottom: 0px;">
        <strong>Backward Pass:</strong> After updating the consequent parameters in the forward pass, the prediction error $E = \frac{1}{2}(y_{pred} - y_{true})^2$ is calculated. The consequent parameters are now kept fixed and the error is propagated backward through the network (backpropagation). Using the <strong>Gradient Descent</strong> algorithm, the premise parameters ($c$ and $\sigma$) are updated:
        <br>
        $$ c_{new} = c_{old} - \eta \frac{\partial E}{\partial c} \quad , \quad \sigma_{new} = \sigma_{old} - \eta \frac{\partial E}{\partial \sigma} $$
        where $\eta$ is the learning rate, and the derivatives are computed using the chain rule across the network layers.
</li>

</ul>

<p style="text-align: left;">
        In the Python code below, this complex architecture has been designed from scratch in the form of a comprehensive and modular class.
</p>

</div>

```python
# Step 2: Hybrid Learning ANFIS Implementation
import numpy as np

class CustomANFIS:
    def __init__(self, n_inputs=2, n_mf_per_input=2):
        """
        Initialize the ANFIS structure.
        We have 2 inputs (V, T) and 2 Gaussian MFs per input.
        Total rules = 2 * 2 = 4.
        """
        self.n_inputs = n_inputs
        self.n_mf = n_mf_per_input
        self.n_rules = self.n_mf ** self.n_inputs
        
        # Initialize Premise Parameters (Gaussian: c, sigma)
        # Randomly distributed within sensible ranges based on the data
        # For V: range (10, 140). Let's init centers around 40 and 110, sigmas around 30.
        # For T: range (5, 95). Let's init centers around 30 and 70, sigmas around 20.
        self.centers = np.array([[40.0, 110.0], [30.0, 70.0]]) # Shape: (2, 2)
        self.sigmas = np.array([[30.0, 30.0], [20.0, 20.0]])   # Shape: (2, 2)
        
        # Initialize Consequent Parameters (p, q, r for each of the 4 rules)
        # Shape: (n_rules, n_inputs + 1) -> (4, 3)
        self.consequents = np.zeros((self.n_rules, self.n_inputs + 1))
        
    def gaussian_mf(self, x, c, sigma):
        """Calculate Gaussian membership degree."""
        # Add small epsilon to sigma to avoid division by zero
        return np.exp(-0.5 * ((x - c) / (sigma + 1e-8))**2)
        
    def forward_pass(self, X):
        """
        Executes Layers 1 to 3 of the ANFIS architecture.
        Returns the normalized firing strengths (w_bar).
        """
        n_samples = X.shape[0]
        
        # Layer 1: Fuzzification
        # mu shape: (n_samples, n_inputs, n_mf) -> (500, 2, 2)
        mu = np.zeros((n_samples, self.n_inputs, self.n_mf))
        for i in range(self.n_inputs):
            for j in range(self.n_mf):
                mu[:, i, j] = self.gaussian_mf(X[:, i], self.centers[i, j], self.sigmas[i, j])
                
        # Layer 2: Rule Firing Strengths (T-norm: product)
        # We have 4 rules. Rule 1: V_low & T_low, Rule 2: V_low & T_high, etc.
        w = np.zeros((n_samples, self.n_rules))
        w[:, 0] = mu[:, 0, 0] * mu[:, 1, 0] # MF0 of Input 0 * MF0 of Input 1
        w[:, 1] = mu[:, 0, 0] * mu[:, 1, 1] # MF0 of Input 0 * MF1 of Input 1
        w[:, 2] = mu[:, 0, 1] * mu[:, 1, 0] # MF1 of Input 0 * MF0 of Input 1
        w[:, 3] = mu[:, 0, 1] * mu[:, 1, 1] # MF1 of Input 0 * MF1 of Input 1
        
        # Layer 3: Normalization
        w_sum = np.sum(w, axis=1, keepdims=True) + 1e-8 # Avoid div by zero
        w_bar = w / w_sum # Shape: (500, 4)
        
        return mu, w, w_bar
        
    def update_consequents_LSE(self, X, y, w_bar):
        """
        Forward Pass (Part 2): Optimize consequent parameters (p,q,r) using 
        Least Squares Estimation (LSE).
        """
        n_samples = X.shape[0]
        
        # Create Design Matrix A
        # For each rule i, the terms are: w_bar_i * V, w_bar_i * T, w_bar_i * 1
        A = np.zeros((n_samples, self.n_rules * (self.n_inputs + 1)))
        
        # Add bias term (1) to inputs
        X_extended = np.hstack((X, np.ones((n_samples, 1)))) # Shape: (500, 3)
        
        for i in range(self.n_rules):
            start_idx = i * (self.n_inputs + 1)
            end_idx = start_idx + (self.n_inputs + 1)
            # Broadcasting: multiply w_bar column by the extended inputs
            A[:, start_idx:end_idx] = w_bar[:, i:i+1] * X_extended
            
        # LSE Calculation: theta = (A^T A)^-1 A^T y
        # Using pseudo-inverse for numerical stability
        theta = np.linalg.pinv(A.T @ A) @ A.T @ y
        
        # Reshape and store back to consequents array
        self.consequents = theta.reshape((self.n_rules, self.n_inputs + 1))
        
    def predict(self, X, w_bar=None):
        """
        Layer 4 & 5: Compute final output using current consequents.
        """
        if w_bar is None:
            _, _, w_bar = self.forward_pass(X)
            
        n_samples = X.shape[0]
        y_pred = np.zeros(n_samples)
        
        X_extended = np.hstack((X, np.ones((n_samples, 1))))
        
        for i in range(self.n_rules):
            # Evaluate rule consequence: f_i = p_i*V + q_i*T + r_i
            f_i = np.sum(X_extended * self.consequents[i, :], axis=1)
            y_pred += w_bar[:, i] * f_i
            
        return y_pred
        
    def backpropagate_GD(self, X, y_true, y_pred, mu, w, w_bar, learning_rate):
        """
        Backward Pass: Update premise parameters (c, sigma) using Gradient Descent.
        Calculates gradients using the chain rule.
        """
        n_samples = X.shape[0]
        X_extended = np.hstack((X, np.ones((n_samples, 1))))
        
        # Error term derivative w.r.t y_pred
        dE_dy = y_pred - y_true # Shape: (500,)
        
        # Precompute rule outputs f_i
        f = np.zeros((n_samples, self.n_rules))
        for i in range(self.n_rules):
            f[:, i] = np.sum(X_extended * self.consequents[i, :], axis=1)
            
        # w_sum for denominator
        w_sum = np.sum(w, axis=1) + 1e-8
        
        # Gradients matrices
        dc = np.zeros_like(self.centers)
        dsigma = np.zeros_like(self.sigmas)
        
        # Iterate over inputs and their MFs to accumulate gradients
        for i in range(self.n_inputs): # i: 0 (V), 1 (T)
            for j in range(self.n_mf): # j: 0 (MF1), 1 (MF2)
                
                # Chain rule step 1: dE / dw_k
                # Which rules involve MF(i,j)?
                # If i=0 (V): MF0 is in Rule 0,1. MF1 is in Rule 2,3.
                # If i=1 (T): MF0 is in Rule 0,2. MF1 is in Rule 1,3.
                if i == 0 and j == 0: active_rules = [0, 1]
                elif i == 0 and j == 1: active_rules = [2, 3]
                elif i == 1 and j == 0: active_rules = [0, 2]
                elif i == 1 and j == 1: active_rules = [1, 3]
                
                # Accumulate derivative for the specific MF across its active rules
                dE_dmu = np.zeros(n_samples)
                for r in active_rules:
                    # dy / dw_r = (f_r - y_pred) / w_sum
                    dy_dw_r = (f[:, r] - y_pred) / w_sum
                    # dw_r / dmu = w_r / mu (since T-norm is product)
                    dw_r_dmu = w[:, r] / (mu[:, i, j] + 1e-8)
                    
                    dE_dmu += dE_dy * dy_dw_r * dw_r_dmu
                
                # Chain rule step 2: dmu / dc and dmu / dsigma
                x_val = X[:, i]
                c_val = self.centers[i, j]
                sig_val = self.sigmas[i, j]
                
                dmu_dc = mu[:, i, j] * ((x_val - c_val) / (sig_val**2 + 1e-8))
                dmu_dsigma = mu[:, i, j] * ((x_val - c_val)**2 / (sig_val**3 + 1e-8))
                
                # Aggregate gradients over all samples
                grad_c = np.sum(dE_dmu * dmu_dc)
                grad_sigma = np.sum(dE_dmu * dmu_dsigma)
                
                # Apply Gradient Descent
                self.centers[i, j] -= learning_rate * grad_c / n_samples
                self.sigmas[i, j] -= learning_rate * grad_sigma / n_samples
                
    def train(self, X, y, epochs=50, learning_rate=0.01):
        """
        Executes the Hybrid Learning algorithm over multiple epochs.
        """
        history_mse = []
        
        for epoch in range(epochs):
            # Phase 1: Forward Pass (Fixed antecedents, LSE for consequents)
            mu, w, w_bar = self.forward_pass(X)
            self.update_consequents_LSE(X, y, w_bar)
            
            # Phase 2: Backward Pass (Fixed consequents, GD for antecedents)
            y_pred = self.predict(X, w_bar)
            mse = np.mean((y_pred - y)**2)
            history_mse.append(mse)
            
            self.backpropagate_GD(X, y, y_pred, mu, w, w_bar, learning_rate)
            
            if (epoch+1) % 10 == 0 or epoch == 0:
                print(f"Epoch {epoch+1:02d}/{epochs} - Train MSE: {mse:.4f}")
                
        return history_mse

# Let's instantiate and train the custom ANFIS on our dataset
print("Starting ANFIS Hybrid Training...")
anfis_model = CustomANFIS(n_inputs=2, n_mf_per_input=2)
# Train for 50 epochs with a small learning rate (to maintain stability)
training_history = anfis_model.train(X_data, y_data, epochs=50, learning_rate=0.005)

# Plot Training Curve
plt.figure(figsize=(8, 4))
plt.plot(training_history, color='blue', linewidth=2, marker='o', markersize=4)
plt.title('ANFIS Training Learning Curve (MSE vs Epochs)', fontsize=12)
plt.xlabel('Epoch', fontsize=10)
plt.ylabel('Mean Squared Error (MSE)', fontsize=10)
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

<h3 style="text-align: left; margin-bottom: 10px;">Step 3: Extraction and Engineering Interpretation of Fuzzy Rules</h3>

<p style="text-align: left;">
        After the hybrid training process is complete, our system has learned how to interpret the input space and assign an appropriate equation to each region. Here we have two input variables:
</p>

<ul style="padding-left: 20px; text-align: left;">

<li style="margin-bottom: 10px;">
        Speed $V$: with membership functions $V_{Low}$ (centered around a low value) and $V_{High}$ (centered around a high value).
</li>

<li style="margin-bottom: 10px;">
        Throttle position $T$: with membership functions $T_{Low}$ and $T_{High}$.
</li>

</ul>

<p style="text-align: left;">
        The consequent part is of the form $Fuel = pV + qT + r$.
</p>

<h4 style="text-align: left;">Engineering Interpretation of the Coefficients:</h4>

<ul style="padding-left: 20px; text-align: left;">

<li style="margin-bottom: 10px;">
        The coefficient $p$ represents the <strong>sensitivity of fuel consumption to changes in speed</strong> in that specific region. If it is positive and large, it means that increasing speed in that operating regime significantly raises fuel consumption (the effect of air resistance at high speeds).
</li>

<li style="margin-bottom: 10px;">
        The coefficient $q$ represents the <strong>sensitivity to throttle pedal position</strong>. Typically, in acceleration states (high throttle), this coefficient has a significant effect because fuel injection is directly dependent on the throttle valve status.
</li>

<li style="margin-bottom: 0px;">
        The constant term $r$ is a baseline for fuel consumption in that rule.
</li>

</ul>

<p style="text-align: left;">
        In the Python code below, the extracted rules are read directly from the trained model parameters and printed.
</p>

</div>

```python
# Step 3: Rule Extraction and Interpretation

# Retrieve learned parameters
c = anfis_model.centers
sig = anfis_model.sigmas
theta = anfis_model.consequents # Shape: (4, 3)

# Helper function to categorize learned centers as 'Low' or 'High'
# By comparing the two centers for each input
v_labels = ["Low" if c[0,0] < c[0,1] else "High", "High" if c[0,0] < c[0,1] else "Low"]
t_labels = ["Low" if c[1,0] < c[1,1] else "High", "High" if c[1,0] < c[1,1] else "Low"]

print("="*60)
print("📚 EXTRACTED FUZZY RULES (TSK First-Order)")
print("="*60)
print(f"Input 1 (Velocity V) MFs : MF1={v_labels[0]} (c={c[0,0]:.1f}), MF2={v_labels[1]} (c={c[0,1]:.1f})")
print(f"Input 2 (Throttle T) MFs : MF1={t_labels[0]} (c={c[1,0]:.1f}), MF2={t_labels[1]} (c={c[1,1]:.1f})")
print("-" * 60)

# Rules logic based on our T-norm combinations in the forward_pass
combinations = [
    (0, 0), # Rule 1: V_MF1 AND T_MF1
    (0, 1), # Rule 2: V_MF1 AND T_MF2
    (1, 0), # Rule 3: V_MF2 AND T_MF1
    (1, 1)  # Rule 4: V_MF2 AND T_MF2
]

for i, (idx_v, idx_t) in enumerate(combinations):
    v_cond = v_labels[idx_v]
    t_cond = t_labels[idx_t]
    
    p = theta[i, 0] # Coefficient for V
    q = theta[i, 1] # Coefficient for T
    r = theta[i, 2] # Constant
    
    rule_text = f"Rule {i+1}: IF Speed is {v_cond} AND Throttle is {t_cond}, "
    rule_text += f"THEN Fuel = ({p:+.4f})*V + ({q:+.4f})*T + ({r:+.4f})"
    print(rule_text)
    
print("="*60)
print("🔍 ENGINEERING INTERPRETATION:")
print("- Notice the coefficients (p and q). A larger positive 'q' in rules where Throttle is 'High'")
print("  indicates a more aggressive injection of fuel during heavy acceleration.")
print("- The mathematical combinations dynamically blend these 4 linear equations based on the")
print("  driver's real-time input (V, T) to create a smooth, highly accurate non-linear output surface.")
```
